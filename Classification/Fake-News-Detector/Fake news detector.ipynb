```python
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import  RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
True_df=pd.read_csv("archive/True.csv")
Fake_df=pd.read_csv("archive/Fake.csv")
```

```python
True_df.head()
```

```python
Fake_df.head()
```

```python
True_df["Label"]=1
Fake_df["Label"]=0
```

```python
True_df.head()
```

```python
Fake_df.head()
```

```python
dataset=pd.concat([Fake_df,True_df],ignore_index=True)
```

```python
dataset
```

```python
data=dataset.sample(frac=1).reset_index(drop=True)
```

```python
data
```

```python
X=data[["title","text","subject","date"]]
Y=data["Label"]
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
vectorizer=TfidfVectorizer(stop_words="english",max_df=0.7)
X_train_tfidf = vectorizer.fit_transform(X_train["text"])
X_test_tfidf = vectorizer.transform(X_test["text"])
```

```python
print(X_train_tfidf.shape) 
print(y_train.shape)  
```

```python
model=RandomForestClassifier()
model.fit(X_train_tfidf, y_train)
```

```python
y_pred = model.predict(X_test_tfidf)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
print( confusion_matrix(y_test, y_pred))
```
